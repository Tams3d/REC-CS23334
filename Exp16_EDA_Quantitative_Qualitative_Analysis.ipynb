```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

try:
    tips = sns.load_dataset("tips")
except Exception:
    import numpy as np

    n = 244
    tips = pd.DataFrame(
        {
            "total_bill": np.random.normal(20, 8, n).clip(3, 60),
            "tip": np.random.normal(3, 1.5, n).clip(1, 10),
            "sex": np.random.choice(["Male", "Female"], n),
            "smoker": np.random.choice(["Yes", "No"], n),
            "day": np.random.choice(["Thur", "Fri", "Sat", "Sun"], n),
            "time": np.random.choice(["Lunch", "Dinner"], n),
            "size": np.random.randint(1, 6, n),
        }
    )
print(tips.head())
```

```python
sns.displot(tips["total_bill"], kde=True)
plt.show()
```

```python
sns.jointplot(x="tip", y="total_bill", data=tips)
plt.show()
```

```python
sns.jointplot(x="tip", y="total_bill", data=tips, kind="reg")
plt.show()
```

```python
sns.pairplot(tips, hue="time")
plt.show()
```

```python
print(tips.corr(numeric_only=True))
sns.heatmap(tips.corr(numeric_only=True), annot=True)
plt.show()
```

```python
sns.boxplot(y=tips["total_bill"])
plt.show()
```

```python
sns.countplot(x="day", data=tips)
plt.show()
sns.countplot(x="sex", data=tips)
plt.show()
```

```python
tips["sex"].value_counts().plot(kind="pie", autopct="%1.1f%%")
plt.show()
tips["sex"].value_counts().plot(kind="bar")
plt.show()
```
