```python
import numpy as np
import pandas as pd
```

```python
df = pd.DataFrame(
    {
        "Country": [
            "France",
            "Spain",
            "Germany",
            "Spain",
            "Germany",
            "France",
            "Spain",
            "France",
            None,
            "France",
        ],
        "Age": [44, 27, 30, 38, 40, 35, None, 48, None, 37],
        "Salary": [72000, 48000, 54000, 61000, None, 58000, 52000, 79000, 83000, 67000],
        "Purchased": ["No", "Yes", "No", "No", "Yes", "Yes", "No", "Yes", "No", "Yes"],
    }
)
print(df)
print(df.info())
```

```python
df["Country"] = df["Country"].fillna(df["Country"].mode()[0])
df["Age"] = df["Age"].fillna(df["Age"].median())
df["Salary"] = df["Salary"].fillna(round(df["Salary"].mean()))
print(df)
```

```python
print(pd.get_dummies(df["Country"]))
```

```python
updated_dataset = pd.concat(
    [pd.get_dummies(df["Country"]), df.iloc[:, [1, 2, 3]]], axis=1
)
updated_dataset["Purchased"].replace(["No", "Yes"], [0, 1], inplace=True)
print(updated_dataset)
```
