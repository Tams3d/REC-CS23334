```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_iris

iris = load_iris(as_frame=True)
data = iris.frame
data.rename(
    columns={
        "sepal length (cm)": "SepalLengthCm",
        "sepal width (cm)": "SepalWidthCm",
        "petal length (cm)": "PetalLengthCm",
        "petal width (cm)": "PetalWidthCm",
        "target": "Id",
    },
    inplace=True,
)
data["variety"] = data["Id"].map(
    {0: "Iris-setosa", 1: "Iris-versicolor", 2: "Iris-virginica"}
)
```

```python
print(data.head())
print(data.info())
print(data.describe())
print(data["variety"].value_counts())
```

```python
sns.countplot(x="variety", data=data)
plt.show()
```

```python
sns.scatterplot(x="SepalLengthCm", y="SepalWidthCm", hue="variety", data=data)
plt.show()
```

```python
sns.scatterplot(x="PetalLengthCm", y="PetalWidthCm", hue="variety", data=data)
plt.show()
```

```python
sns.pairplot(data, hue="variety")
plt.show()
```

```python
for col in ["PetalLengthCm", "PetalWidthCm", "SepalLengthCm", "SepalWidthCm"]:
    sns.FacetGrid(data, hue="variety", height=5).map(sns.histplot, col).add_legend()
    plt.show()
```
