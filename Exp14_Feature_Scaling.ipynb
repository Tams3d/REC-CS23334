```python
import numpy as np
import pandas as pd
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, StandardScaler, MinMaxScaler
```

```python
df = pd.DataFrame(
    {
        "Country": [
            "France",
            "Spain",
            "Germany",
            "Spain",
            "Germany",
            "France",
            "Spain",
            "France",
            None,
            "France",
        ],
        "Age": [44, 27, 30, 38, 40, 35, None, 48, 50, 37],
        "Salary": [72000, 48000, 54000, 61000, None, 58000, 52000, 79000, 83000, 67000],
        "Purchased": ["No", "Yes", "No", "No", "Yes", "Yes", "No", "Yes", "No", "Yes"],
    }
)
df["Country"] = df["Country"].fillna(df["Country"].mode()[0])
```

```python
features = df.iloc[:, :-1].values
label = df.iloc[:, -1].values
age_imp = SimpleImputer(strategy="mean")
sal_imp = SimpleImputer(strategy="mean")
features[:, [1]] = age_imp.fit_transform(features[:, [1]])
features[:, [2]] = sal_imp.fit_transform(features[:, [2]])
print(features)
```

```python
oh = OneHotEncoder(sparse_output=False)
Country = oh.fit_transform(features[:, [0]])
print(Country)
```

```python
final_set = np.concatenate((Country, features[:, [1, 2]].astype(float)), axis=1)
print(final_set)
```

```python
sc = StandardScaler()
print(sc.fit_transform(final_set))
```

```python
mms = MinMaxScaler(feature_range=(0, 1))
print(mms.fit_transform(final_set))
```
