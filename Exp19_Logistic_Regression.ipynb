```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report
```

```python
np.random.seed(0)
n = 400
age = np.random.randint(18, 60, n)
salary = np.random.randint(15000, 150000, n)
purchased = ((age + salary / 5000 + np.random.normal(0, 3, n)) > 45).astype(int)
df = pd.DataFrame({"Age": age, "EstimatedSalary": salary, "Purchased": purchased})
print(df.head())
```

```python
features = df.iloc[:, [0, 1]].values
label = df.iloc[:, 2].values
x_train, x_test, y_train, y_test = train_test_split(
    features, label, test_size=0.2, random_state=0
)
```

```python
finalModel = LogisticRegression(max_iter=1000)
finalModel.fit(x_train, y_train)
print(finalModel.score(x_train, y_train))
print(finalModel.score(x_test, y_test))
```

```python
print(classification_report(label, finalModel.predict(features)))
```
