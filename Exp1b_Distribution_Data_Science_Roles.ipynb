```python
import matplotlib.pyplot as plt

roles = [
    "Data Analyst",
    "Data Engineer",
    "Data Scientist",
    "ML Engineer",
    "Business Analyst",
]
counts = [300, 500, 450, 200, 150]
```

```python
plt.bar(roles, counts)
plt.title("Distribution of Data Science Roles")
plt.xlabel("Role")
plt.ylabel("Count")
plt.xticks(rotation=15)
plt.show()
```

```python
plt.pie(counts, labels=roles, autopct="%1.1f%%")
plt.title("Distribution of Data Science Roles")
plt.show()
```
