```python
import numpy as np
```

```python
array = np.random.randint(1, 100, 9)
print(array)
print(np.sqrt(array))
print(array.ndim)
```

```python
new_array = array.reshape(3, 3)
print(new_array)
print(new_array.ndim)
```

```python
print(new_array.ravel())
print(new_array[2, 1:3])
print(new_array[1:2, 1:3])
print(new_array[0:2, 0:1])
print(new_array[0:3, 0:1])
print(new_array[1:3])
```
